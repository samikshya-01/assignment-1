# 1. Setup

```python
import numpy as np
import time

np.random.seed(7)
cities = np.array(["Kathmandu", "Pokhara", "Biratnagar", "Lalitpur", "Birgunj"])
days = np.array([f"D{i:02d}" for i in range(1, 15)])

temps = np.random.randint(14, 36, size=(5, 14)).astype(float)                 # °C, 5 cities × 14 days
humidity = np.random.randint(40, 95, size=(5, 14)).astype(float)              # %

temps[2, 5] = np.nan          # sensor outage
temps[4, 11] = np.nan
```

# 2. B1

```python
# ---------- B1: Array anatomy and reshaping ----------
print("B1 shape:", temps.shape, "| ndim:", temps.ndim, "| dtype:", temps.dtype, "| size:", temps.size)
a = np.arange(70)
r1 = a.reshape(5, 14)
r2 = a.reshape(5, -1)
print("B1 explicit vs -1 identical:", np.array_equal(r1, r2))
print("B1 flattened back, shape:", r1.ravel().shape)
##What -1 means?
#Suppose you have a single row of 70 objects and you want to divide them into 5 rows which means each row gets 14 objects
# because 70/5 = 14 and that is what -1 do like i will give you the rows and you figure out the rest.
# Just like how I wrote (5, -1) in the code and numpy gave me 14.

##Why reshape doesn't copy?
#Reshape doesn't copy because in memory, data is stored in a continuous line and reshaping is just a way of changing how
# we view it not change the actual data or copy it. And it is more efficient because it takes less memory.
```

# 3. B2

```python
# ---------- B2: Indexing and slicing ----------
print("B2a Pokhara day 3:", temps[1, 2], "shape:", temps[1, 2].shape)
print("B2b every city, last day:", temps[:, -1], "shape:", temps[:, -1].shape)
print("B2c Biratnagar fortnight:", temps[2], "shape:", temps[2].shape)
print("B2d first 3 cities, days 4-8:\n", temps[:3, 3:8], "shape:", temps[:3, 3:8].shape)
print("B2e Lalitpur every other day:", temps[3, ::2], "shape:", temps[3, ::2].shape)
print("B2f Kathmandu reversed:", temps[0, ::-1], "shape:", temps[0, ::-1].shape)
```

# 4. B3

```python
# ---------- B3: Boolean masks ----------
hot = temps > 30
print("B3a readings above 30 per city:", hot.sum(axis=1))
print("B3b days Kathmandu exceeded 30:", days[hot[0]])
##What is a mask and why temps[temps > 30] can't keep the 2D shape?
#A mask is a True/False grid which returns the only two possible outcomes and temps goes flat because Kathmandu has 1 hot
# day while Birgunj has 6 so numpy cannot keep it as rows since every row should be same length within the grid which
# is not possible with two different values so it cannot keep the 2D shapeand gives a simple list.
```

# 5. B4

```python
# ---------- B4: Views vs copies ----------
demo = temps.copy()          # work on a copy so the real temps stays untouched for B5-B10
view = demo[0, :3]
view[0] = 999
print("B4 original after changing the VIEW:", demo[0, :3])      # 999 appears

demo = temps.copy()
cp = demo[0, :3].copy()
cp[0] = 999
print("B4 original after changing the COPY:", demo[0, :3])      # unchanged
##Why NumPy defaults to a view, and what bug this causes if you forget?
#Numpy defaults to a view because it saves memory and you can change the original data without having to copy it and
# if you forget that it is a view and accidentally change the original data, then it causes bugs in the code.
# So, I only recommend this when you actually know what to change and how will it work and if not then just use a copy to
# avoid errors.
```

# 6. B5

```python
# ---------- B5: Vectorization vs a Python loop ----------
f_loop = np.empty_like(temps)
for i in range(temps.shape[0]):              # the only loop allowed in Part B
    for j in range(temps.shape[1]):
        f_loop[i, j] = temps[i, j] * 9 / 5 + 32
f_vec = temps * 9 / 5 + 32
print("B5 loop and vectorized agree:", np.allclose(f_loop, f_vec, equal_nan=True))

big = np.random.rand(1_000_000) * 30
start = time.perf_counter()
out = np.empty_like(big)
for i in range(big.size):
    out[i] = big[i] * 9 / 5 + 32
loop_time = time.perf_counter() - start

start = time.perf_counter()
out_vec = big * 9 / 5 + 32
vec_time = time.perf_counter() - start
print(f"B5 loop: {loop_time:.4f} s | vectorized: {vec_time:.4f} s | ~{loop_time / vec_time:.0f}x faster")
##Where does the loop's extra time go?
#The loop's extra time goes into checking each element with specific instructions which causes it to take longer than Numpy
# because numpy uses fast C code to perform the same thing in a single operation which makes it faster.
```

# 7. B6

```python
# ---------- B6: Broadcasting ----------
city_avg = np.nanmean(temps, axis=1, keepdims=True)
anomaly = temps - city_avg
print("B6a city averages shape:", city_avg.shape, "| anomaly shape:", anomaly.shape)

correction = np.linspace(-0.5, 0.5, 14)
corrected = temps + correction
print("B6b correction shape:", correction.shape, "| corrected shape:", corrected.shape)

try:
    temps + np.array([1, 2, 3, 4, 5])
except ValueError as e:
    print("B6c error:", e)
##Why does (5,) fail but (14,) work?
#Numpy matches arrays from the right side and the right side of temps is 14 days so (14,) has 14 values so it fits and gets
# added to every city while (5,) has only 5 values, which doesn't match the 14 days, so numpy gives an error.
```

# 8. B7

```python
# ---------- B7: Matrix multiplication ----------
quantities = np.array([2, 3, 1])
prices = np.array([120, 90, 250])
print("B7a quantities * prices:", quantities * prices)
print("B7a quantities @ prices:", quantities @ prices)
##What did * and @ return, and why do they differ?
# * returns the element-wise multiplication of the two arrays while @ returns the dot product of the two arrays.
# They differ because * multiplies each corresponding element while @ sums the products of corresponding elements.

features = np.column_stack([np.nanmean(temps, axis=0), np.nanmean(humidity, axis=0)])
weights = np.array([0.6, -0.1])
comfort = features @ weights
print("B7b features:", features.shape, "| weights:", weights.shape, "| comfort:", comfort.shape)
print("B7b comfort score per day:", np.round(comfort, 2))
##Why did the inner dimensions have to match?
#Inner dimensions have to match because the number and weight of the features must match for the dot product like temperature and
# humidity have 2 features and weights also have 2 features so they can be multiplied but if weights had 3 features
# then it would not work.
#* returned [240 270 250] (2*120, 3*90, 1*250) and @ returned 760 (240 + 270 + 250).
```
