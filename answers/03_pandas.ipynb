# 11. C10: Derived columns (3 pts)
Add passed (marks >= 40), and grade using nested np.where: 85+ → A, 70–84 → B, 40–69 → C, below 40 → F. Print the overall pass rate as a percentage by taking the mean of passed cast to int, and print grade.value_counts(). Deal with the impossible marks values you found in C2, and say in a comment what you did with them and why.

```python
# ---------- C10: Derived columns ----------
print("C10 impossible marks (> 100):", (df["marks"] > 100).sum())
df = df[df["marks"] <= 100].copy()
print("C10 shape after removing them:", df.shape)

df["passed"] = df["marks"] >= 40
df["grade"] = np.where(df["marks"] >= 85, "A",
              np.where(df["marks"] >= 70, "B",
              np.where(df["marks"] >= 40, "C", "F")))

print("C10 pass rate:", round(df["passed"].astype(int).mean() * 100, 2), "%")
print(df["grade"].value_counts())
##What I did with the impossible marks and why?
#I removed the impossible marks because they are errors and since marks cannot be greater than 100, its better to remove them than
# replace with random value which may cause greater error in the data.
```

# 12. C11: Types and memory (3 pts)
Convert subject and city to the category dtype. Print memory_usage(deep=True) for those two columns before and after, in bytes, and the percentage saved. In a comment, explain what makes category cheaper, and when it would not help.

```python
# ---------- C11: Types and memory ----------
before = df[["subject", "city"]].memory_usage(deep=True)[["subject", "city"]]
df["subject"] = df["subject"].astype("category")
df["city"] = df["city"].astype("category")
after = df[["subject", "city"]].memory_usage(deep=True)[["subject", "city"]]

print("C11 bytes BEFORE:\n", before)
print("C11 bytes AFTER:\n", after)
print("C11 % saved:\n", ((before - after) / before * 100).round(1))
##Why category is cheaper, and when it would not help?
#Category is cheaper because it store the value only once and then store the index of the value for other which saves memory but it
# wouldn't help when every value is different because it will take more memory.
```

# 13. C12: Aggregation and a summary table (3 pts)
On the cleaned data, produce: (a) mean, max and count of marks per subject using groupby().agg(); (b) mean marks by city and subject; (c) a pivot_table of mean marks with cities as rows and subjects as columns; (d) mean marks per exam_month, sorted chronologically. Export (a) to outputs/summary.csv, and state in one sentence what the month trend shows.

```python
# ---------- C12: Aggregation and summary ----------
summary = df.groupby("subject", observed=True)["marks"].agg(["mean", "max", "count"]).round(2)
print("C12a per subject:"); display(summary)

print("C12b mean marks by city and subject:")
display(df.groupby(["city", "subject"], observed=True)["marks"].mean().round(2))

print("C12c pivot table:")
display(df.pivot_table(values="marks", index="city", columns="subject", aggfunc="mean", observed=True).round(1))

print("C12d mean marks per month:")
display(df.groupby("exam_month")["marks"].mean().sort_index().round(2))

summary.to_csv("../outputs/summary.csv")
print("C12 saved outputs/summary.csv")
##What the month trend shows?
#The month trend shows that the mean marks are highest in January and lowest in February but the difference is small.
```
