# 2. C1

```python
# ---------- C1: Series and DataFrame by hand ----------
marks = pd.Series([78, 85, 62, 91, 55], index=["Aarav", "Bibek", "Chandni", "Deepika", "Esha"])
print("C1 Series index:", marks.index.tolist())
print("C1 Series values:", marks.values)

df_a = pd.DataFrame({
    "name": ["Aarav", "Bibek", "Chandni"],
    "subject": ["Statistics", "Databases", "Machine Learning"],
    "marks": [78, 85, 62],
})
df_b = pd.DataFrame([
    {"name": "Aarav", "subject": "Statistics", "marks": 78},
    {"name": "Bibek", "subject": "Databases", "marks": 85},
    {"name": "Chandni", "subject": "Machine Learning", "marks": 62},
])
print("C1 dict-of-lists equals list-of-dicts:", df_a.equals(df_b))
print("C1 shape:", df_a.shape)
print("C1 columns:", df_a.columns.tolist())
print("C1 index:", df_a.index.tolist())
```

```python
# ---------- C2: Load and inspect ----------
df = pd.read_csv("../data/scores_raw.csv")
print("C2 head:"); display(df.head())
print("C2 tail(3):"); display(df.tail(3))
print("C2 shape:", df.shape)
df.info()
print("C2 describe:"); display(df.describe())
print("C2 describe(include='object'):"); display(df.describe(include="object"))
##Three problems visible in the output
#1. marks max is 150 in describe(), which is wrong because the max marks receivable is only 100.
#2. Some values are missing like 4 rows have no marks and 10 rows have no attendance (info() shows 122/126 and 116/126).
#3. city has 24 unique values in describe(include="object") but there are only 6 cities in the data.
```

# 4. C3

```python
# ---------- C3: Selecting rows and columns ----------
print("C3 df['marks'] type:", type(df["marks"]))
print("C3 df[['marks']] type:", type(df[["marks"]]))

loc_part = df.loc[0:4, ["name", "subject", "marks"]]
iloc_part = df.iloc[0:5, 1:4]
display(loc_part)
print("C3 loc rows:", len(loc_part))
display(iloc_part)
print("C3 iloc rows:", len(iloc_part))
##Difference between df["marks"] and df[["marks"]].
#df["marks"] returns a series while df[["marks"]] returns a dataframe.

##Why loc and iloc treat the end point differently?
#loc is label based so it includes end point while iloc uses index so it excludes the end point.
```

# 5. C4

```python
# ---------- C4: Boolean filtering ----------
c4a = df[(df["marks"] > 80) & (df["subject"] == "Python Programming")]
print("C4a Python Programming with marks > 80:", len(c4a), "rows")
display(c4a)

c4b = df[df["city"].isin(["Kathmandu", "Pokhara"])]
print("C4b Kathmandu or Pokhara (before cleaning):", len(c4b), "rows")
# NOTE: this under-counts because city is still messy ("KATHMANDU", "  Pokhara ", ...). Fixed in C5.

c4c = df.query("marks > 80 and subject == 'Python Programming'")
print("C4c query result equals C4a:", c4a.equals(c4c))
##Why and/or fail inside brackets but are required inside query()
#and/or fail inside brackets because they can only check one True/False, not a whole column but they are required inside query() because
# it is a string and pandas checks for and/or not & and |.
```
